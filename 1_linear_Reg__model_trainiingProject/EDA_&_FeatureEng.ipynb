### EDA aur Feature Engineering Kya Hota Hai? (Short Intro)

#### 1. Exploratory Data Analysis (EDA)

* **Kya hai?** Data ka X-Ray / Post-Mortem karna. Modeling se pehle data ki nature, patterns aur anomalies ko visuals (charts) aur numbers se samajhna.
* **Kaise karte hain?** Graphs (Histograms, Boxplots, Heatmaps, Scatter plots) aur Statistical Summary (`describe()`, `corr()`) ka use karke.
* **Kyun karte hain?** Taaki pata chale features me kya relation hai, outliers hain ya nahi, aur konsa feature fire/FWI ko sabse zyada effect kar raha hai.

#### 2. Feature Engineering

* **Kya hai?** Raw data ko aisi format me badalna jisse ML Model aasaani se seekh sake aur better predictions de sake.
* **Kaise karte hain?** New features create karke, unwanted features remove karke, values encode karke, ya scales normalise karke.
* **Kyun karte hain?** Machine learning models pure raw numbers ya un-cleaned text ko samajhte nahi hain, unhe crisp aur relevant features chahiye hote hain.

---

### Algerian Forest Fires Project Me Sir Aage Kya-Kya Karenge?

Iss project ke aage ke steps me sir ye main cheezein cover karenge:

#### Phase A: Feature Engineering (Data Preprocessing)

1. **Date Column Handling (Row-level cleaning):**
* Dataset me `day`, `month`, `year` separate columns hain. Sir check karenge ki kya inki zaroorat hai ya inhe drop karna sahi rahega.


2. **Category Encoding (Target Feature Fix):**
* `Classes` column me text values hain (`'fire'`, `'not fire'`, `'fire '` with trailing spaces).
* Sir pehle `.str.strip()` karke spaces hatayenge.
* Uske baad isse numerical format me convert karenge: `fire` $\rightarrow$ `1`, `not fire` $\rightarrow$ `0`.


3. **Multicollinearity Removal (VIF / Correlation Filter):**
* FWI system ke indexes (`FFMC`, `DMC`, `DC`, `ISI`, `BUI`) ek doosre se strongly related hote hain.
* Sir high correlation wale features ko spot karke duplicate information hatayenge taaki model overfit na ho.



#### Phase B: Exploratory Data Analysis (EDA)

1. **Univariate Analysis (Ek-ek feature ka behavior):**
* **Histograms / KDE Plots:** Dekhne ke liye ki `Temperature`, `RH`, `Rain` ka distribution kesa hai (skewed hai ya normal/bell curve).
* **Pie Chart / Count Plot:** Check karne ke liye ki `fire` vs `not fire` ke kitne counts hain (dataset balanced hai ya nahi).


2. **Bivariate / Multivariate Analysis (Features ke aapas ka relation):**
* **Heatmap (Correlation Matrix):** Sir `sns.heatmap(df.corr(), annot=True)` banayenge taaki direct pata chal sake ki kis feature ka `FWI` ya `Classes` par sabse zyada positive/negative impact hai.
* **Box Plots:** `Temperature` vs `Classes` (Fire) check karne ke liye ki high temperature par aag lagne ke kitne chances hote hain aur dataset me Outliers kahan hain.
* **Region-wise Comparison:** `Bejaia` vs `Sidi Bel-abbes` me se kis region me zyada aag lagi thi.

Ab aap lecture me dekhenge ki sir ek-ek karke `matplotlib` aur `seaborn` se ye saare charts plot karenge aur correlation heatmap se decision lenge!

````python
# lets start EDA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cleaned CSV file ko read karein
dataset = pd.read_csv('cleaned_forest_fire_dataset.csv')

# Data check karein
dataset.head()

df_copy = dataset.drop(['day', 'month', 'year'], axis=1)
````

````python
df_copy
````

````python
df_copy['Classes'] = np.where(df_copy['Classes']== 'not fire',0,1)
````

````python
df_copy
````

````python
df_copy['Classes'].value_counts()
````

````python
## plot density plot for all features

plt.style.use('seaborn-v0_8')\


df_copy.hist(bins=50, figsize=(20,15))
plt.show()
````

````python
percentage = df_copy['Classes'].value_counts(normalize=True) * 100

plt.pie(percentage, labels=["fire","not fire"], autopct='%1.1f %%', colors=['#ff9999','#66b3ff'])
plt.title("percentage of classes")
plt.show()
````

2. Box Plot (Outliers Dhoondhne ke Liye)
Linear Regression ke liye yeh graph bohot zaroori hai!

Kyun banate hain: Linear Regression outliers ke samne bohot kamzor hota hai (ek bada outlier pure regression line ko kharab kar deta hai).


````python
# Rain, DMC, DC, ISI jaise features me outliers check karein
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_copy[['Rain', 'FFMC', 'DMC', 'ISI']])
plt.title("Outlier Analysis")
plt.show()
````

Feature vs Class Box Plot (Bivariate Analysis)
Kyun banate hain: Yeh dekhne ke liye ki kis feature ke badhne/ghatne se aag lagti hai.

````python
sns.boxplot(x='Classes', y='Temperature', data=df_copy)
plt.title("Temperature vs Classes (0=Not Fire, 1=Fire)")
plt.show()
````

Correlation Heatmap (🔥 SABSE IMPORTANT GRAPH)
Agar aap Linear Regression model banane ja rahe hain, toh is graph ke bina aap aage nahi badh sakte!

Kyun banate hain:
Yeh dekhne ke liye ki kaunsa feature Target (FWI) ke sath sabse zyada strongly related hai.
Multicollinearity check karne ke liye: Agar do independent features aapas mein bohot strongly correlated hain (jaise DMC aur BUI ka correlation > 0.85 hota hai), toh Linear Regression confuse ho jaata hai aur galat weights seekh leta hai. Hume unme se ek ko drop karna padta hai.


````python
plt.figure(figsize=(15,10))

sns.heatmap(df_copy.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("correlation heatmap")
plt.show()
````
