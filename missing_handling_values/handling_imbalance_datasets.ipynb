```python
import numpy as np
import pandas as pd 

# set the random seed for reporducibility
np.random.seed(123)

#create a df with two classes

n_sample = 1000
class_0_ratio = 0.9
n_class_0 = int(n_sample * class_0_ratio)
n_class_1 = n_sample - n_class_0
```

```python
n_class_0,n_class_1
```

```python
### create my df with imbalanced dataset

# Class 0 ke liye features create karna (Majority Class - 900 records)
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})


# Class 1 ke liye features create karna (Minority Class - 100 records)
class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
# Dono classes ko ek saath concat karke main DataFrame banana
df = pd.concat([class_0, class_1]).reset_index(drop=True)

# Check target class distribution
df['target'].value_counts()
```

# upsampling


```python

df_minority = df[df['target'] == 1]

df_majority = df[df['target'] == 0]
```

```python
from sklearn.utils import resample

df_minority_upsample = resample(df_minority, replace=True, # sample with replacement
        n_samples=len(df_majority),
        random_state=42
)
```

```python
df_minority_upsample.head()
```

```python
df_unsample = pd.concat([df_majority, df_minority_upsample])
```

```python
df_unsample['target'].value_counts()
```
