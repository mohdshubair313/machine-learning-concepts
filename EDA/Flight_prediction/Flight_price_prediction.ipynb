## 3️⃣ Date, Month, Year ko Separate Kyu Aur Kaise Karte Hain? (Feature Engineering)

### ❓ Kyu alag karte hain?
Machine Learning model ko agar aap direct `'11-02-2022'` doge, to model ko ye sirf ek bekaar text lagega, wo isse kuch seekh nahi payega.

Lekin agar aap ise todkar alag-alag columns bana doge:
* **`Date` (Day):** `11`
* **`Month`:** `2` (February)
* **`Year`:** `2022`

To Model aasani se pattern pakad lega (jaise: *Festivals/Holidays wale month me ticket mehengi hoti hai, ya weekend wali date par price badh jata hai*).

---

### 🛠️ Hume Kaise Karna Hoga? (Code)

Aapke `df['date']` me format hai: `'11-02-2022'` (`DD-MM-YYYY`).

Ise alag karne ke **2 simple tareeqe** hain:

#### 🔹 Tarika 1: String Split Method (`.str.split`)
```python
# 1. Date (Day) nikaalna
df['Date'] = df['date'].str.split('-').str[0].astype(int)

# 2. Month nikaalna
df['Month'] = df['date'].str.split('-').str[1].astype(int)

# 3. Year nikaalna
df['Year'] = df['date'].str.split('-').str[2].astype(int)

# 4. Ab purane 'date' text column ko delete kar do
df.drop('date', axis=1, inplace=True)

df[['Date', 'Month', 'Year']].head()
```

#### 🔹 Tarika 2: Pandas DateTime Method (Professional Way)
```python
# 'date' column ko datetime me convert karein
df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y')

# Date, Month, Year extract karein
df['Date'] = df['date'].dt.day
df['Month'] = df['date'].dt.month
df['Year'] = df['date'].dt.year

# Original date column drop karein
df.drop('date', axis=1, inplace=True)
```

---

### 💡 Ek Important Tip aapke dataset ke liye (`price` column):
Aapke `df['price']` me comma laga hua hai (jaise `'25,612'`), isliye wo `str` ban gaya hai. Use number banane ke liye yeh zaroor karna:

```python
# Comma hatakar Integer/Float me convert karein
df['price'] = df['price'].str.replace(',', '').astype(int)
```

````python
#feature engineering 

# 1. Date (Day) nikaalna
df['Date'] = df['date'].str.split('-').str[0].astype(int)
# 2. Month nikaalna
df['Month'] = df['date'].str.split('-').str[1].astype(int)
# 3. Year nikaalna
df['Year'] = df['date'].str.split('-').str[2].astype(int)
# 4. Ab purane 'date' text column ko delete kar do
df.drop('date', axis=1, inplace=True)
df[['Date', 'Month', 'Year']].head()
````

````python
df['arrival_hour'] = df['arr_time'].str.split(':').str[0].astype(int)
df['arrival_min'] = df['arr_time'].str.split(':').str[1].astype(int)
````

````python
# now split the dep_time also into two parts hours, min

df['dep_hour'] = df['dep_time'].str.split(':').str[0].astype(int)
df['dep_min'] = df['dep_time'].str.split(':').str[1].astype(int)
````

````python
df['stop'] = df['stop'].str.split().str[0]

df['stop'].unique()
````

````python
import numpy as np
df['Total_stops'] = df['stop'].map({'non-stop':0, '1-stop':1, '2+-stop':2, np.nan:1})
````

````python
df[['stop', 'Total_stops']]
````

````python
# wants the duration in hours or minutes totally

df['duration_hrs']= df['time_taken'].str.split(' ').str[0].str.replace('h','').astype(int)

# extract the minutes section
df['duration_mins'] = df['time_taken'].str.split(' ').str[1].str.replace('m', '').astype(int)

# total minutes calculating 
df['total_duration_in_mins'] = (df['duration_hrs'] * 60) + df['duration_mins']

# 4. Ab purane text wale 'time_taken' column ko drop kar dein
df.drop('time_taken', axis=1, inplace=True)
````
