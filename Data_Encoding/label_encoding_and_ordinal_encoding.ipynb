### 💡 OneHotEncoder vs LabelEncoder me Main Difference:
| Feature | OneHotEncoder | LabelEncoder |
| :--- | :--- | :--- |
| **Input Shape** | 2D Table: `df[['color']]` | 1D Series: `df['color']` |
| **Output** | Naye binary columns (`0` aur `1`) | Same column me numbers (`0, 1, 2, ...`) |
| **Kab Use Karein?** | Nominal features (Red, Green, Blue) jahan koi order nahi hota | Target Column ($y$) ya binary categories (Yes/No, Male/Female) |

Bhai, yeh Scikit-Learn ka ek bohot famous **Design Architecture & Purpose** ka concept hai. Aao samjhte hain ki dono me yeh farq kyu rakha gaya hai:

---

## 🎯 1. Asli Wajah: Dono Ka Maqsad (Purpose) Alag Hai!

Machine Learning me data ke do hisse hote hain:
1. **$X$ (Input Features / Independent Variables)** $\rightarrow$ Yeh hamesha **2D Table** hoti hai (Multiple rows & columns).
2. **$y$ (Target / Output Variable)** $\rightarrow$ Yeh hamesha **1D List/Vector** hota hai (Single column - jaise `Yes/No`, `Spam/Not Spam`).

```text
       Input Features (X) [2D]            Target (y) [1D]
  +--------+--------+----------------+       +----------+
  |  Age   | Gender |      City      |       | Purchased|
  +--------+--------+----------------+       +----------+
  |   25   |  Male  |     Delhi      |  -->  |   Yes    |
  |   30   | Female |     Mumbai     |  -->  |    No    |
  +--------+--------+----------------+       +----------+
```

---

## 🔹 Reason 1: `LabelEncoder` Target ($y$) ke liye banaya gaya hai (1D)

* **Design:** `LabelEncoder` ko sirf **Output/Target column ($y$)** ko encode karne ke liye banaya gaya tha.
* Target hamesha ek single column (1D Series) hota hai, jaise `['Cat', 'Dog', 'Cat']` $\rightarrow$ `[0, 1, 0]`.
* Iska kaam koi naye columns banana nahi hai, bas usi 1D line me text ko number banana hai.
* Isliye yeh hamesha **1D Array `(N,)`** leta hai: `df['target']`.

---

## 🔹 Reason 2: `OneHotEncoder` Input Features ($X$) ke liye banaya gaya hai (2D)

* **Design:** `OneHotEncoder` ko **Input Features ($X$)** ke liye banaya gaya hai.
* Input me ek sath **kayi saare categorical columns** ho sakte hain (e.g., `City`, `Gender`, `Color`).
* `OneHotEncoder` ek hi baar me multiple columns ko encode karke ek badi **2D Matrix** bana deta hai:
  ```python
  # Ek sath 2 columns ko encode kar sakta hai
  encoder.fit_transform(df[['City', 'Gender']])
  ```
* Isliye iska input aur output dono hamesha **2D `(N_rows, N_features)`** hote hain.

---

## 💡 Ek Interesting Baat: Agar Input Features ($X$) ko numbers me badalna ho to?

Agar aap input feature $X$ ko `0, 1, 2` me badalna chahte ho aur 2D me kaam karna chahte ho, to Scikit-Learn ne uske liye alag tool banaya hai: **`OrdinalEncoder`**!

```python
from sklearn.preprocessing import OrdinalEncoder

# Yeh 2D input leta hai aur numbers (0, 1, 2) deta hai
ord_encoder = OrdinalEncoder()
ord_encoder.fit_transform(df[['color', 'city']])
```

---

## 📊 Summary Table:

| Tool | Kiske Liye Banaya Gaya? | Input Shape | Kaam Kya Karta Hai? |
| :--- | :--- | :--- | :--- |
| **`LabelEncoder`** | **Target variable ($y$)** | **1D** (`df['target']`) | 1 column ke text ko numbers me badalta hai. |
| **`OneHotEncoder`** | **Input features ($X$)** | **2D** (`df[['feature1', 'feature2']]`) | Multiple columns ke liye naye binary (0/1) columns banata hai. |
| **`OrdinalEncoder`** | **Input features ($X$)** | **2D** (`df[['feature1', 'feature2']]`) | Multiple columns ke text ko numbers (0, 1, 2) me badalta hai. |

# Ordinal Encoding

````python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
# DataFrame jisme natural order (rank) ho
df_edu = pd.DataFrame({
    'qualification': ['High School', 'Bachelors', 'Masters', 'PhD', 'Bachelors', 'Masters']
})
df_edu
````

````python
from sklearn.preprocessing import OrdinalEncoder

# Order define karna (Chhote se Bada)
encoder = OrdinalEncoder(categories=[['High School', 'Bachelors', 'Masters', 'PhD']])

# 2. Transform karein (2D DataFrame [[]] pass karein)
df_edu['qualification_encoded'] = encoder.fit_transform(df_edu[['qualification']])

df_edu
````
