# 3.Target Encoding

```python
import pandas as pd

df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 300, 250, 180, 320]
})
```

```python
mean_price = df.groupby('city')['price'].mean().to_dict()

mean_price
```

```python
df['city_encoded'] = df['city'].map(mean_price)

df
```

```python
import seaborn as sns
import pandas as pd
df = sns.load_dataset('tips')

df[['time', 'total_bill']].head()
```

```python
# Lunch aur Dinner ka average bill calculate karke dictionary banayein
mean_time_dict = df.groupby('time')['total_bill'].mean().to_dict()

mean_time_dict

# (Yani Lunch ka average bill lagbhag ₹17.16 hai aur Dinner ka ₹20.79).
```

```python
# map() function use karke naya column banaye

df['time_encoded'] = df['time'].map(mean_time_dict)

df[['time', 'total_bill', 'time_encoded']].head()
```

```python
# Order nikalna mean ke hisab se
ordered_labels = df.groupby('time')['total_bill'].mean().sort_values().index

# Rank dictionary: {'Lunch': 1, 'Dinner': 2}
rank_dict = {k: i for i, k in enumerate(ordered_labels, 1)}

# Map karna
df['time_rank_encoded'] = df['time'].map(rank_dict)

df[['time', 'total_bill', 'time_encoded', 'time_rank_encoded']].head()
```
